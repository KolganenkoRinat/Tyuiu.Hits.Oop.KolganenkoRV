<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Invoice в C#, который будет представлять информацию о
фактурах за поставленные товары или оказанные услуги. На основе этого класса
разработать 2-3 производных класса, демонстрирующих принципы наследования и
полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и
методы, а также переопределены некоторые методы базового класса для
демонстрации полиморфизма.



<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
// 1. Вспомогательный класс для позиции в счете
public class LineItem
{
    public string Name { get; set; }
    public decimal Price { get; set; }

    public LineItem(string name, decimal price)
    {
        Name = name;
        Price = price;
    }
}

// 2. БАЗОВЫЙ КЛАСС
public class Invoice
{
    public string InvoiceNumber { get; set; }
    public DateTime IssueDate { get; set; }
    public decimal TotalAmount { get; protected set; } // protected — доступно внутри наследников
    
    protected List<LineItem> Items = new List<LineItem>();

    public Invoice(string number, DateTime date)
    {
        InvoiceNumber = number;
        IssueDate = date;
    }

    // Виртуальный метод для расчета суммы (Полиморфизм)
    public virtual decimal CalculateTotal()
    {
        TotalAmount = Items.Sum(item => item.Price);
        return TotalAmount;
    }

    // Виртуальный метод добавления позиции
    public virtual void AddLine(LineItem lineItem)
    {
        Items.Add(lineItem);
        Console.WriteLine($"Добавлена позиция: {lineItem.Name} ({lineItem.Price} руб.)");
    }

    // Виртуальный метод удаления позиции
    public virtual void RemoveLine(LineItem lineItem)
    {
        Items.Remove(lineItem);
        Console.WriteLine($"Удалена позиция: {lineItem.Name}");
    }
}

// 3. ТОВАРНАЯ ФАКТУРА (Наследование)
public class GoodsInvoice : Invoice
{
    public DateTime SupplyDate { get; set; }

    public GoodsInvoice(string number, DateTime date, DateTime supplyDate) 
        : base(number, date)
    {
        SupplyDate = supplyDate;
    }

    // Переопределение метода (Полиморфизм)
    public override void AddLine(LineItem lineItem)
    {
        base.AddLine(lineItem);
        Console.WriteLine($"--> Товар будет поставлен: {SupplyDate.ToShortDateString()}");
    }
}

// 4. УСЛУГОВАЯ ФАКТУРА (Наследование)
public class ServiceInvoice : Invoice
{
    public DateTime ServiceDate { get; set; }

    public ServiceInvoice(string number, DateTime date, DateTime serviceDate) 
        : base(number, date)
    {
        ServiceDate = serviceDate;
    }

    // Переопределение метода (Полиморфизм)
    public override void RemoveLine(LineItem lineItem)
    {
        base.RemoveLine(lineItem);
        Console.WriteLine("--> Причина удаления: Услуга была отменена клиентом.");
    }
}

// 5. КОМБИНИРОВАННАЯ ФАКТУРА (Наследование)
public class CombinedInvoice : Invoice
{
    public bool ReturnAllowed { get; set; }

    public CombinedInvoice(string number, DateTime date, bool returnAllowed) 
        : base(number, date)
    {
        ReturnAllowed = returnAllowed;
    }

    // Переопределение метода расчета суммы (Полиморфизм)
    public override decimal CalculateTotal()
    {
        decimal baseTotal = base.CalculateTotal();
        
        if (ReturnAllowed)
        {
            TotalAmount = baseTotal + (baseTotal * 0.05m); // Сбор 5% за возможность возврата
            Console.WriteLine("(Включен сбор за возможность возврата 5%)");
        }
        return TotalAmount;
    }
}

// ==========================================
// ИСПОЛНЯЕМЫЙ КОД (Проверка работы классов)
// ==========================================

Console.WriteLine("=== Работа с товарной фактурой ===");
GoodsInvoice goods = new GoodsInvoice("Т-001", DateTime.Now, DateTime.Now.AddDays(3));
goods.AddLine(new LineItem("Ноутбук", 50000));

Console.WriteLine("\n=== Работа с услуговой фактурой ===");
ServiceInvoice service = new ServiceInvoice("У-001", DateTime.Now, DateTime.Now);
LineItem consulting = new LineItem("Консультация", 5000);
service.AddLine(consulting);
service.RemoveLine(consulting);

Console.WriteLine("\n=== Работа с комбинированной фактурой ===");
CombinedInvoice combined = new CombinedInvoice("К-001", DateTime.Now, true);
combined.AddLine(new LineItem("Смартфон", 30000));
combined.AddLine(new LineItem("Настройка ПО", 2000));

decimal total = combined.CalculateTotal();
Console.WriteLine($"Итоговая сумма комбинированного счета: {total} руб.");
```
